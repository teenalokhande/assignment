 2. Compare and contrast np.mean() and np.average() functions in NumPy. When would you use one over the
other?

--> np.mean() calculates the arithmetic mean of array elements along a specified axis. np.average() does the same but allows for weighted averages if weights are provided. Use np.average() when weighting is required; otherwise, np.mean() is simpler and often faster.

3) Describe the methods for reversing a NumPy array along different axes. Provide examples for 1D and 2D
arrays.

--> For 1D arrays, reverse with array[::-1]. In 2D arrays, np.flip(array, axis=0) reverses rows, while np.flip(array, axis=1) reverses columns. Example:

1D: [1, 2, 3][::-1] → [3, 2, 1]
2D: np.flip([[1,2], [3,4]], axis=0) → [[3,4], [1,2]]

4. How can you determine the data type of elements in a NumPy array? Discuss the importance of data types
in memory management and performance

-->  Use array.dtype to find the data type of elements. Data types affect memory use and processing speed; smaller types like int8 save memory, while larger types like float64 are more precise but memory-intensive.

9. Discuss the functionality of the array_split() method in NumPy. How does it handle uneven splits?

--> array_split() divides an array into multiple sub-arrays along a specified axis. If the array doesn’t divide evenly, array_split() creates smaller sub-arrays in such a way that each resulting piece has nearly equal lengths, which is useful in scenarios where chunked data processing is needed.



10. Explain the concepts of vectorization and broadcasting in NumPy. How do they contribute to efficient array
operations?

--> Vectorization allows applying operations to entire arrays at once without explicit loops, making code both faster and easier to read. Broadcasting lets NumPy perform operations on arrays with different shapes by extending them automatically to compatible shapes, simplifying code and speeding up computation without additional memory allocation.
