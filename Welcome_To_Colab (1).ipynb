1. What are the five key concepts of Object-Oriented Programming (OOP)?

--> The five key concepts of OOP are:

Class: A blueprint or template for creating objects, which defines a set of attributes and methods.

Object: An instance of a class. Each object can have different attributes but shares the same class structure.

Encapsulation: Bundling the data (attributes) and the methods (functions) that operate on the data within a single unit (class), and restricting access to some of the object's components.

Inheritance: A way to form new classes using existing classes. This allows for reusability and extension of code.

Polymorphism: The ability of different objects to respond to the same method in different ways. It allows for one interface to be used for multiple data types.

3. Explain the difference between instance methods and class methods. Provide an example of each.

--> Instance methods: These methods operate on an instance of the class (object) and can access instance-specific data using self. They are used for behaviors that pertain to a particular object.

Class methods: These methods operate on the class itself rather than instances of the class. They use cls as their first parameter and are called using the class name. Class methods are often used to create factory methods.

```
class MyClass:
    def __init__(self, value):
        self.value = value

    # instance method
    def show_value(self):
        return self.value

    # class method
    @classmethod
    def from_string(cls, string_value):
        return cls(int(string_value))

obj = MyClass(10)
print(obj.show_value())

obj2 = MyClass.from_string("20")
print(obj2.show_value())
```

4. How does Python implement method overloading? Give an example.

--> Python does not support traditional method overloading (defining multiple methods with the same name but different parameters). However, method overloading can be simulated by using default arguments or by using *args and **kwargs to accept a variable number of arguments.

```
class Example:
    def add(self, a, b, c=0):
        return a + b + c

obj = Example()
print(obj.add(1, 2))
print(obj.add(1, 2, 3))
```

 5. What are the three types of access modifiers in Python? How are they denoted?

 --> Python has three types of access modifiers:

Public (no underscore): Members are accessible from outside the class. Example: self.attribute.

Protected (_): Members are intended for internal use, but can still be accessed from outside the class. It's more of a convention. Example: self._attribute.

Private (__): Members are not accessible directly outside the class, and the name is "name-mangled" by Python to make it harder to access. Example: self.__attribute.


```
class Test:
    def __init__(self):
        self.public = "I am public"
        self._protected = "I am protected"
        self.__private = "I am private"

    def show_private(self):
        return self.__private

obj = Test()
print(obj.public)
print(obj._protected)
print(obj.show_private())


```

6. Describe the five types of inheritance in Python. Provide a simple example of multiple inheritance.

--> The five types of inheritance are:

Single Inheritance: A subclass inherits from one parent class.

Multiple Inheritance: A subclass inherits from more than one parent class.

Multilevel Inheritance: A subclass is derived from another subclass.

Hierarchical Inheritance: Multiple subclasses inherit from the same parent class.

Hybrid Inheritance: A combination of two or more types of inheritance.

```
class A:
    def method_a(self):
        return "Method from A"

class B:
    def method_b(self):
        return "Method from B"

class C(A, B):
    pass

obj = C()
print(obj.method_a())
print(obj.method_b())
```

 7. What is the Method Resolution Order (MRO) in Python? How can you retrieve it programmatically?

 --> MRO is the order in which Python looks for a method in a hierarchy of classes during inheritance. It follows the C3 linearization algorithm, ensuring that a method is searched for in the current class, then parent classes, in a left-to-right depth-first manner.

```
class A: pass
class B(A): pass
class C(A): pass
class D(B, C): pass

print(D.mro())
```

 8. Create an abstract base class Shape with an abstract method area(). Then create two subclasses

 -->

```
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

# Example usage:
circle = Circle(5)
rectangle = Rectangle(4, 6)
print(circle.area())
print(rectangle.area())
```

 9. Demonstrate polymorphism by creating a function that can work with different shape objects to calculate and print their areas.

 -->

```
from abc import ABC, abstractmethod #Importing necessary modules for abstract classes

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

def print_area(shape):
    print(f"The area is {shape.area()}")

circle = Circle(5)
rectangle = Rectangle(4, 6)
print_area(circle)
print_area(rectangle)
```

10. Implement encapsulation in a BankAccount class with private attributes for balance and account_number. Include methods for deposit, withdrawal, and balance inquiry.

-->


```
class BankAccount:
    def __init__(self, account_number, initial_balance=0):
        self.__account_number = account_number
        self.__balance = initial_balance

    def deposit(self, amount):
        self.__balance += amount

    def withdraw(self, amount):
        if amount > self.__balance:
            return "Insufficient funds"
        self.__balance -= amount

    def get_balance(self):
        return self.__balance

account = BankAccount("123456")
account.deposit(100)
print(account.get_balance())
account.withdraw(50)
print(account.get_balance())
```

 11. Write a class that overrides the __str__ and __add__ magic methods. What will these methods allow you to do?

 -->

```
class MyNumber:
    def __init__(self, value):
        self.value = value

    def __str__(self):
        return f"MyNumber({self.value})"

    def __add__(self, other):
        return MyNumber(self.value + other.value)

num1 = MyNumber(10)
num2 = MyNumber(20)
result = num1 + num2
print(result)
```

12. Create a decorator that measures and prints the execution time of a function.

-->

```
import time

def time_decorator(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"Execution time: {end_time - start_time} seconds")
        return result
    return wrapper

@time_decorator
def example_function():
    time.sleep(2)
    return "Function executed"

example_function()
```

13. Explain the concept of the Diamond Problem in multiple inheritance. How does Python resolve it?

--> The Diamond Problem occurs in multiple inheritance when a class inherits from two classes that both inherit from a common base class, creating ambiguity about which class's method should be inherited.
Python resolves it using the MRO (Method Resolution Order), which ensures that each class is only called once, and the method resolution follows a left-to-right depth-first approach.

```
class A:
    def greet(self):
        return "Hello from A"

class B(A): pass
class C(A): pass
class D(B, C): pass

obj = D()
print(obj.greet())
```

14. Write a class method that keeps track of the number of instances created from a class.

-->

```
class MyClass:
    instance_count = 0

    def __init__(self):
        MyClass.instance_count += 1

    @classmethod
    def get_instance_count(cls):
        return cls.instance_count

obj1 = MyClass()
obj2 = MyClass()
print(MyClass.get_instance_count())
```

14. Write a class method that keeps track of the number of instances created from a class.

-->

```
class Calendar:
    @staticmethod
    def is_leap_year(year):
        return year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)

print(Calendar.is_leap_year(2020))
print(Calendar.is_leap_year(2021))
```
